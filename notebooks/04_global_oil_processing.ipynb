```python
import pandas as pd

brent = pd.read_csv(
    "d:/HK3 - năm 3/Đồ Án/Code/vietnam-transportation-cpi-forecasting/data/raw/brent_crude_oil_monthly.csv"
)

wti = pd.read_csv(
    "d:/HK3 - năm 3/Đồ Án/Code/vietnam-transportation-cpi-forecasting/data/raw/wti_crude_oil_monthly.csv"
)
```

```python
brent.head()
```

```python
wti.head()
```

```python
brent["Date"] = pd.to_datetime(brent["Date"])
wti["Date"] = pd.to_datetime(wti["Date"])
print(brent.dtypes)
print(wti.dtypes)
```

```python
brent = brent[
    (brent["Date"] >= "2011-09-01") &
    (brent["Date"] <= "2024-12-31")
].copy()

wti = wti[
    (wti["Date"] >= "2011-09-01") &
    (wti["Date"] <= "2024-12-31")
].copy()
print("Brent:", brent["Date"].min(), "->", brent["Date"].max())
print("WTI:", wti["Date"].min(), "->", wti["Date"].max())
```

```python
brent["MonthYear"] = brent["Date"].dt.to_period("M")
wti["MonthYear"] = wti["Date"].dt.to_period("M")
print(brent[["Date", "MonthYear", "Price"]].head())
print(wti[["Date", "MonthYear", "Price"]].head())
```

```python
brent = brent.rename(columns={"Price": "Brent"})
wti = wti.rename(columns={"Price": "WTI"})
print(brent.head())
print(wti.head())
```

```python
oil_global = brent[["MonthYear", "Brent"]].merge(
    wti[["MonthYear", "WTI"]],
    on="MonthYear",
    how="inner"
)
oil_global.head()
```

```python
print(oil_global.shape)
```

```python
oil_global.isna().sum()
```

```python
oil_global[
    (oil_global["Brent"] < 0) |
    (oil_global["WTI"] < 0)
]
```

```python
brent_monthly = oil_global[
    ["MonthYear", "Brent"]
].copy()

wti_monthly = oil_global[
    ["MonthYear", "WTI"]
].copy()

brent_monthly.to_csv(
    "d:/HK3 - năm 3/Đồ Án/Code/vietnam-transportation-cpi-forecasting/data/interim/brent_monthly.csv",
    index=False,
    encoding="utf-8-sig"
)

wti_monthly.to_csv(
    "d:/HK3 - năm 3/Đồ Án/Code/vietnam-transportation-cpi-forecasting/data/interim/wti_monthly.csv",
    index=False,
    encoding="utf-8-sig"
)
```
