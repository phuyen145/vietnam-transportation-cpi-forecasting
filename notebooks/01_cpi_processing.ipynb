```python
raw = pd.read_excel(file_path, header=None)

raw.head()
```

```python
df = raw.copy()

df.columns = [
    "Year",
    "M01", "M02", "M03", "M04",
    "M05", "M06", "M07", "M08",
    "M09", "M10", "M11", "M12"
]

df.head()
```

```python
df = df.drop(index=0).reset_index(drop=True)
df.head()
```

```python
df = df.apply(
    pd.to_numeric,
    errors="coerce"
)
df.head()
```

```python
df.isna().sum()
```

```python
df.shape
```

```python
df["Year"] = df["Year"].astype(int)
df["Year"].tolist()
df.head()
```

```python
cpi_long = df.melt(
    id_vars="Year",
    var_name="Month",
    value_name="CPI"
)
cpi_long.head()
```

```python
cpi_long["MonthYear"] = pd.to_datetime(
    cpi_long["Year"].astype(int).astype(str) + "-" +
    cpi_long["Month"].str.replace("M", "", regex=False),
    format="%Y-%m"
).dt.to_period("M")
cpi_long.head(15)
```

```python
cpi_long = cpi_long.sort_values("MonthYear").reset_index(drop=True)
cpi_long.head(15)
```

```python
print(cpi_long.shape)
```

```python
cpi_long = cpi_long.set_index("MonthYear")
cpi_long.head(15)
```

```python
cpi_long.dtypes
```

```python
print(cpi_long.head())
print(cpi_long.tail())
```

```python
cpi_long = cpi_long.drop(columns=["Year", "Month"])
cpi_long.head()
```

```python
cpi_long["CPI"] = (cpi_long["CPI"] - 100).round(2)
cpi_long.head()
```

```python
cpi_long.to_csv(
    "d:/HK3 - năm 3/Đồ Án/Code/vietnam-transportation-cpi-forecasting/data/interim/cpi_transport_monthly.csv",
    encoding="utf-8-sig"
)
```
