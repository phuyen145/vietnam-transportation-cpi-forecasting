```python
import unicodedata

fuel_1["mat_hang_clean"] = (
    fuel_1["mat_hang"]
    .astype(str)
    .apply(lambda x: unicodedata.normalize("NFKC", x))
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)
print(fuel_1["mat_hang_clean"].value_counts().to_string())
```
